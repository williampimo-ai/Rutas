```python
# ==========================================
# Sistema de Ubicación y Rutas basado en Reglas
# Autor: William Pinzón
# ==========================================

print("--- BIENVENIDO A MI DE NAVEGACIÓN INTELIGENTE ---")
print("Por favor configura las condiciones de tu viaje:\n")

# ------------------------------------------
# ENTRADAS DEL USUARIO (Menús interactivos)
# ------------------------------------------

punto_partida = "Casa"
punto_llegada = "Universidad"

# Elegir medio de transporte
print("¿En qué te vas a movilizar?")
print("1. Caminando")
print("2. Bicicleta")
print("3. Moto")
print("4. Automóvil")
print("5. Bus")
opcion_transporte = input("Digita el número de tu opción (1-5): ")

# Mapeamos la opción numérica a texto limpio
if opcion_transporte == "1":
    medio_transporte = "caminando"
elif opcion_transporte == "2":
    medio_transporte = "bicicleta"
elif opcion_transporte == "3":
    medio_transporte = "moto"
elif opcion_transporte == "4":
    medio_transporte = "automovil"
elif opcion_transporte == "5":
    medio_transporte = "bus"
else:
    print("Opción inválida, por defecto nos iremos en bus.")
    medio_transporte = "bus"

# Elegir clima
print("\n¿Cómo está el clima hoy?")
print("1. Despejado / Soleado")
print("2. Lluvioso")
print("3. Nublado")
opcion_clima = input("Digita el número de tu opción (1-3): ")

if opcion_clima == "2":
    clima_actual = "lluvia"
elif opcion_clima == "3":
    clima_actual = "nublado"
else:
    clima_actual = "despejado"

# Preguntar el estado de la hora / tráfico
print("\n¿Cómo está la movilidad a esta hora?")
print("1. Tráfico pesado / Hora pico")
print("2. Tráfico normal / Fluido")
print("3. Todo despejado (sin tráfico)")
opcion_movilidad = input("Digita el número de tu opción (1-3): ")

if opcion_movilidad == "1":
    estado_movilidad = "hora_pico"
    nivel_trafico = "alto"
    es_hora_pico = True
elif opcion_movilidad == "2":
    estado_movilidad = "normal"
    nivel_trafico = "medio"
    es_hora_pico = False
else:
    estado_movilidad = "sin_trafico"
    nivel_trafico = "bajo"
    es_hora_pico = False

# Convertimos el booleano a "Sí" o "No" para que quede natural
texto_hora_pico = "Sí" if es_hora_pico else "No"

# Mostramos el resumen limpio con el cambio aplicado
print("\n" + "-"*45)
print(f"Ruta: De {punto_partida} a {punto_llegada}")
print(f"Selección: Transporte ({medio_transporte}) | Clima ({clima_actual}) | Tráfico ({nivel_trafico}) | Hora pico: {texto_hora_pico}")
print("-"*45 + "\n")

# Tiempo base por defecto en minutos según el medio elegido
if medio_transporte == "caminando":
    tiempo_estimado = 45
elif medio_transporte == "bicicleta":
    tiempo_estimado = 30
elif medio_transporte in ["moto", "automovil"]:
    tiempo_estimado = 20
else: # bus
    tiempo_estimado = 35


# ------------------------------------------
# APLICACIÓN DE REGLAS BASADAS EN EL CLIMA
# ------------------------------------------

if clima_actual == "lluvia" and medio_transporte in ["caminando", "bicicleta"]:
    print(">> [Regla Clima]: Está lloviendo y vas movilizandote al aire libre, posiblemente te vas a mojar, mejor vete con capa o cambia de transporte.")
    tiempo_estimado += 20

elif clima_actual == "lluvia" and medio_transporte in ["moto", "automovil", "bus"]:
    print(">> [Regla Clima]: Lluvia detectada. Las vías se ponen lentas y resbalosas, hay que manejar con más precaución.")
    tiempo_estimado += 15

elif clima_actual == "nublado":
    print(">> [Regla Clima]: El día está nublado, puede que llueva en el camino, por ahora puedes viajar tranquilo, pero atengo al cambio de clima.")


# ------------------------------------------
# APLICACIÓN DE REGLAS DE TRÁFICO Y MOVILIDAD
# ------------------------------------------

if estado_movilidad == "hora_pico" and nivel_trafico == "alto":

    if medio_transporte == "automovil":
        print(">> [Regla Tráfico]: Trancón pesado en hora pico y en carro... Dificultades en el tráfico. Sugiero tomar ruta alterna evitando vias principales.")
        tiempo_estimado += 35

    elif medio_transporte == "moto":
        print(">> [Regla Tráfico]: Hora pico pesada, pero como vas en moto puedes sortear un poco los carros. Igual ve con cuidado.")
        tiempo_estimado += 10

    elif medio_transporte == "bus":
        print(">> [Regla Tráfico]: El bus va a demorarse más de lo normal por la congestión vehicular.")
        tiempo_estimado += 25

    elif medio_transporte == "bicicleta":
        print(">> [Regla Tráfico]: Hay tráfico de carros, pero en bici puedes usar la ciclorruta y te ahorras gran parte.")
        tiempo_estimado += 5

elif estado_movilidad == "sin_trafico":
    print(">> [Regla Tráfico]: Vías completamente despejadas, el viaje va a ser rápido.")
    tiempo_estimado -= 5


# ------------------------------------------
# RESULTADO FINAL Y RECOMENDACIÓN
# ------------------------------------------

print("\n" + "="*40)
print("          REPORTE FINAL DE RUTA")
print("="*40)
print(f"Medio seleccionado: {medio_transporte.capitalize()}")
print(f"Tiempo estimado total: {tiempo_estimado} minutos")

if tiempo_estimado > 50:
    print("💡 Consejo: Sal con bastante tiempo de anticipación o evalúa otra opción de movilidad.")
else:
    print("💡 Consejo: Tiempos aceptables, ¡Vamonos!")
```
