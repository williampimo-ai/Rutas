```python
# ==========================================
# Sistema de Ubicación y Rutas basado en Reglas
# Autor: William Pinzón
# Proyecto académico - Lógica difusa / Reglas
# ==========================================

print("--- BIEVENIDO AL SISTEMA DE NAVEGACIÓN INTELIGENTE ---")
print("Por favor configuremos las condiciones de su viaje de hoy:\n")

# ------------------------------------------
# ENTRADAS DEL USUARIO (Menús interactivos)
# ------------------------------------------

punto_partida = "Casa"
punto_llegada = "Universidad"

# Elegir medio de transporte
print("¿En qué se va a movilizar?")
print("1. Caminando")
print("2. Bicicleta")
print("3. Moto")
print("4. Automóvil")
print("5. Bus")
opcion_transporte = input("Digite el número de su opción (1-5): ")

# Mapeamos la opción numérica a texto limpio
if opcion_transporte == "1":
    medio_transporte = "caminando"
elif opcion_transporte == "2":
    medio_transporte = "bicicleta"
elif opcion_transporte == "3":
    medio_transporte = "moto"
elif opcion_transporte == "4":
    medio_transporte = "automovil"
elif opcion_transporte == "5":
    medio_transporte = "bus"
else:
    print("Opción inválida, por defecto nos iremos en bus.")
    medio_transporte = "bus"

# Elegir clima
print("\n¿Cómo está el clima?")
print("1. Despejado / Soleado")
print("2. Lluvia")
print("3. Neblina")
opcion_clima = input("Digite el número de su opción (1-3): ")

if opcion_clima == "2":
    clima_actual = "lluvia"
elif opcion_clima == "3":
    clima_actual = "neblina"
else:
    clima_actual = "despejado"

# Elegir tráfico
print("\n¿Cómo ve el nivel de tráfico?")
print("1. Bajo")
print("2. Medio")
print("3. Alto")
opcion_trafico = input("Digite el número de su opción (1-3): ")

if opcion_trafico == "1":
    nivel_trafico = "bajo"
elif opcion_trafico == "2":
    nivel_trafico = "medio"
else:
    nivel_trafico = "alto"

# Preguntar si es hora pico
opcion_pico = input("\n¿Es hora pico? (s/n): ").lower()
hora_pico = True if opcion_pico == 's' else False

# Mostramos el resumen de lo que escogió el usuario
print("\n" + "-"*45)
print(f"Ruta: De {punto_partida} a {punto_llegada}")
print(f"Selección: Transporte ({medio_transporte}) | Clima ({clima_actual}) | Tráfico ({nivel_trafico}) | Hora pico: {hora_pico}")
print("-"*45 + "\n")

# Tiempo base por defecto en minutos (varía según el medio elegido inicialmente)
if medio_transporte == "caminando":
    tiempo_estimado = 45
elif medio_transporte == "bicicleta":
    tiempo_estimado = 30
elif medio_transporte in ["moto", "automovil"]:
    tiempo_estimado = 20
else: # bus
    tiempo_estimado = 35


# ------------------------------------------
# APLICACIÓN DE REGLAS BASADAS EN EL CLIMA
# ------------------------------------------

# Si llueve y vamos en bici o caminando, la cosa se pone fea y toca sumar tiempo
if clima_actual == "lluvia" and medio_transporte in ["caminando", "bicicleta"]:
    print(">> [Regla Clima]: Está lloviendo y vas a pie o en bici. Te vas a mojar, mejor vete con capa o cambia de transporte.")
    tiempo_estimado += 20

elif clima_actual == "lluvia" and medio_transporte in ["moto", "automovil", "bus"]:
    print(">> [Regla Clima]: Lluvia detectada. Las vías se ponen lentas y resbalosas, hay que manejar con más precaución.")
    tiempo_estimado += 15

elif clima_actual == "neblina":
    print(">> [Regla Clima]: Hay neblina densa, la visibilidad es baja, toca ir despacio.")
    tiempo_estimado += 10


# ------------------------------------------
# APLICACIÓN DE REGLAS DE TRÁFICO Y HORA PICO
# ------------------------------------------

if hora_pico and nivel_trafico == "alto":

    # Si vas en carro en pleno trancón, es lo peor
    if medio_transporte == "automovil":
        print(">> [Regla Tráfico]: Trancón pesado en hora pico y en carro... Paila, te vas a tragar todo el tráfico. Sugiero tomar ruta alterna por la paralela.")
        tiempo_estimado += 35

    # La moto se defiende mejor filtrando carros
    elif medio_transporte == "moto":
        print(">> [Regla Tráfico]: Hora pico pesada, pero como vas en moto puedes sortear un poco los carros. Igual ve con cuidado.")
        tiempo_estimado += 10

    # El bus sufre por el tráfico general
    elif medio_transporte == "bus":
        print(">> [Regla Tráfico]: El bus va a demorarse más de lo normal por la congestión vehicular.")
        tiempo_estimado += 25

    # La bici a veces se salva si hay ciclo vía
    elif medio_transporte == "bicicleta":
        print(">> [Regla Tráfico]: Hay tráfico de carros, pero en bici puedes usar la ciclorruta y te ahorras gran parte.")
        tiempo_gist = 5

elif nivel_trafico == "bajo":
    print(">> [Regla Tráfico]: Vías despejadas, el viaje va a ser rápido.")
    tiempo_estimado -= 5


# ------------------------------------------
# RESULTADO FINAL Y RECOMENDACIÓN
# ------------------------------------------

print("\n" + "="*40)
print("          REPORTE FINAL DE RUTA")
print("="*40)
print(f"Medio seleccionado: {medio_transporte.capitalize()}")
print(f"Tiempo estimado total: {tiempo_estimado} minutos")

if tiempo_estimado > 50:
    print("💡 Consejo: Sal con bastante tiempo de anticipación o evalúa otra opción de movilidad.")
else:
    print("💡 Consejo: Tiempos aceptables, ¡buen viaje!")
```
